```python
import pandas as pd
import numpy as np
import os

# -----------------------------
# Load top submissions
# -----------------------------
exp09 = pd.read_csv("../submissions/exp09_elasticnet_hyperparameter_search_20260323.csv", header=None)
exp13 = pd.read_csv("../submissions/exp13_pls_final_model_20260323.csv", header=None)
exp03 = pd.read_csv("../submissions/exp03_savgol_deriv1_species_pls12_20260323.csv", header=None)

# -----------------------------
# Assign weights
# -----------------------------
weights = {
    "exp09": 0.5,
    "exp13": 0.3,
    "exp03": 0.2
}

# -----------------------------
# Weighted ensemble
# -----------------------------
ensemble_preds = (
    exp09[1] * weights["exp09"] +
    exp13[1] * weights["exp13"] +
    exp03[1] * weights["exp03"]
)

# -----------------------------
# Create submission
# -----------------------------
submission = pd.DataFrame({
    "sample number": exp09[0],
    "含水率": ensemble_preds
})

# -----------------------------
# Save submission
# -----------------------------
os.makedirs("../submissions", exist_ok=True)
output_file = "../submissions/exp15_weighted_ensemble_20260323.csv"
submission.to_csv(output_file, index=False, header=False)

print("Ensemble submission saved:", output_file)
print("Sample predictions:", submission.head())
```
