```python
import pandas as pd
import numpy as np
import os
import re

# --------------------------------
# Folder containing submissions
# --------------------------------
SUBMISSION_DIR = "../submissions"


# --------------------------------
# Load all submission files
# --------------------------------
def load_submissions():

    print("\nScanning submission folder...\n")

    files = [f for f in os.listdir(SUBMISSION_DIR) if f.endswith(".csv")]

    # extract experiment numbers
    def get_exp_number(name):

        match = re.search(r'exp(\d+)', name)

        if match:
            return int(match.group(1))
        return 9999

    files = sorted(files, key=get_exp_number)

    print("Experiments found:")
    for f in files:
        print("  ", f)

    models = {}

    for f in files:

        path = os.path.join(SUBMISSION_DIR, f)

        df = pd.read_csv(path, header=None)

        preds = df[1].values

        models[f] = preds

    print("\nTotal experiments loaded:", len(models))

    return models


# --------------------------------
# Prediction distribution
# --------------------------------
def prediction_diagnostics(models):

    print("\nPrediction Distribution Diagnostics")
    print("-----------------------------------")

    for name, preds in models.items():

        print("\n", name)

        print("mean:", np.mean(preds))
        print("std :", np.std(preds))
        print("min :", np.min(preds))
        print("max :", np.max(preds))


# --------------------------------
# Model correlation
# --------------------------------
def correlation_diagnostics(models):

    print("\nModel Correlation Matrix")
    print("------------------------")

    df = pd.DataFrame(models)

    corr = df.corr()

    print(corr)


# --------------------------------
# Ensemble simulation
# --------------------------------
def ensemble_simulation(models):

    print("\nSimple Ensemble Simulation")
    print("--------------------------")

    df = pd.DataFrame(models)

    ensemble = df.mean(axis=1)

    print("Ensemble prediction stats")

    print("mean:", ensemble.mean())
    print("std :", ensemble.std())
    print("min :", ensemble.min())
    print("max :", ensemble.max())


# --------------------------------
# Main
# --------------------------------
def main():

    models = load_submissions()

    prediction_diagnostics(models)

    correlation_diagnostics(models)

    ensemble_simulation(models)


if __name__ == "__main__":
    main()
```
